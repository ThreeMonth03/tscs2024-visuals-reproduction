## Regenerate the figures (optional)

Leave the settings unchanged for the demonstration. The demo always uses the public aggregate tables. A blank output directory creates a new temporary folder on each run, so running all cells again will not overwrite earlier results.

The GitHub source keeps outputs empty; Binder builds the saved preview from the same public tables when it prepares the environment.


```python
MODE = "aggregate"  # This demo never reads survey records.
OUTPUT_DIR = ""      # Blank creates a fresh temporary folder on each run.
FONT = ""            # Blank uses the installed default.
```

```python
import sys
from pathlib import Path

for candidate in (Path.cwd(), *Path.cwd().parents):
    if (candidate / "run.py").is_file() and (candidate / "manifest.json").is_file():
        sys.path.insert(0, str(candidate))
        break

from helpers.notebook_workflow import configuration, find_root, run_workflow

root = find_root()
if MODE != "aggregate":
    raise ValueError("Use reproduce.ipynb for survey reestimation.")
settings = configuration({
    "REPRO_MODE": "aggregate",
    "REPRO_OUTPUT_DIR": OUTPUT_DIR,
    "REPRO_FONT": FONT,
})
report = run_workflow(root, settings)
print("Verified public aggregates and redrew nine figures. No respondent records were used.")
```
